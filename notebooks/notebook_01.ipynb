# Wearable Health Analytics — Sample Heart Rate Dataset

This notebook demonstrates the basic data import and summary workflow for heart rate signals using Python and pandas.

## Data Preview and Basic Statistics

We start by loading the sample heart rate dataset and verifying that the structure looks correct.

```python
import pandas as pd

df = pd.read_csv("../data/sample_data.csv")
print(df.head())
print("Rows:", len(df))
print("HR mean:", df["hr"].mean())
```

```python
print("Shape:", df.shape)
print("\nData Types:")
print(df.dtypes)
print("\nMissing Values:")
print(df.isna().sum())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.plot(df["id"], df["hr"], marker="o")
plt.xlabel("Participant ID")
plt.ylabel("Heart Rate (bpm)")
plt.title("Heart Rate by Participant")
plt.grid(True)
plt.show()
```
